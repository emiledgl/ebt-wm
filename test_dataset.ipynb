```python
from dataset import DROIDDataset

dataset = DROIDDataset(
    num_episodes=10,
    tubelets_per_clip=16,
    image_size=384,
    tubelet_size=2,
    camera_keys=["observation.images.wrist_left"],
    normalize=False,
    video_backend="pyav",
)
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt

from matplotlib import animation
from IPython.display import HTML

sample = dataset[random.randint(0, len(dataset) - 1)]
video, proprios, actions = sample

# Interpolate actions to match video frames
num_frames = video.shape[0]
action_indices = np.linspace(0, actions.shape[0] - 1, num_frames)
actions_interp = np.zeros((num_frames, actions.shape[1]))

for i in range(num_frames):
    idx = action_indices[i]
    lower_idx = int(np.floor(idx))
    upper_idx = int(np.ceil(idx))
    
    if lower_idx == upper_idx:
        actions_interp[i] = actions[lower_idx].numpy()
    else:
        # Linear interpolation
        weight = idx - lower_idx
        actions_interp[i] = (1 - weight) * actions[lower_idx].numpy() + weight * actions[upper_idx].numpy()

# Visualize clip with proprioception and actions
fig = plt.figure(figsize=(16, 8))
gs = fig.add_gridspec(2, 2, width_ratios=[2, 1], height_ratios=[1, 1], hspace=0.3, wspace=0.3)

ax_video = fig.add_subplot(gs[:, 0])
ax_proprio = fig.add_subplot(gs[0, 1])
ax_action = fig.add_subplot(gs[1, 1])

ims = []
for i in range(video.shape[0]):
    artists = []
    
    # Video frame
    im = ax_video.imshow(video[i].permute(1, 2, 0).numpy(), animated=True)
    title = ax_video.text(0.5, 1.05, f'Frame {i+1}/{num_frames}', 
                          transform=ax_video.transAxes, ha='center', fontsize=12)
    artists.extend([im, title])
    
    # Proprioception bar chart
    proprio_bars = ax_proprio.bar(range(proprios.shape[1]), proprios[i].numpy(), 
                                   color='steelblue', alpha=0.7)
    ax_proprio.set_ylim([proprios.min().item() - 0.1, proprios.max().item() + 0.1])
    ax_proprio.set_xlabel('Proprio Dimension')
    ax_proprio.set_ylabel('Value')
    ax_proprio.set_title('Proprioception')
    ax_proprio.grid(True, alpha=0.3)
    artists.extend(proprio_bars)
    
    # Action bar chart
    action_bars = ax_action.bar(range(actions.shape[1]), actions_interp[i], 
                                 color='coral', alpha=0.7)
    ax_action.set_ylim([actions_interp.min() - 0.1, actions_interp.max() + 0.1])
    ax_action.set_xlabel('Action Dimension')
    ax_action.set_ylabel('Value')
    ax_action.set_title('Action (Interpolated)')
    ax_action.grid(True, alpha=0.3)
    artists.extend(action_bars)
    
    ims.append(artists)

ax_video.axis('off')

ani = animation.ArtistAnimation(fig, ims, interval=100, blit=True, repeat_delay=1000)
plt.close()
HTML(ani.to_jshtml())
```
